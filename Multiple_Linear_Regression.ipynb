```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.float_format', lambda x: '%.2f' % x)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
df_ = pd.read_csv("/content/advertising.csv")
df = df_.copy()
df.head()
```

```python
df.info()
```

```python
df.describe().T
```

## Model

```python
X = df.drop('sales',axis=1)
y= df[['sales']]
X_train, X_test, y_train, y_test = train_test_split(X, y , test_size=0.25, random_state=20)
y_test.shape[0]
```

```python
y_train.shape[0]
```

```python
reg_model= LinearRegression().fit(X_train ,y_train)
```

```python
# constant (b - bias)
reg_model.intercept_[0]
```

```python
# coefficients (w - weights)
reg_model.coef_[0][0]
```

## Prediction

```python
#MSE Train
#Sales mean = 14.02
y_pred = reg_model.predict(X_train)
np.sqrt(mean_squared_error(y_train, y_pred))
```

```python
#MAE Train
np.sqrt(mean_absolute_error(y_train, y_pred))
```

```python
# R² Train
# R² is interpreted as follows:
# 1.0: The model explains the data perfectly.
# 0.0: The model explains the target variable as if it were a random guess.
reg_model.score(X_train, y_train)
```

```python
# MSE Test
y_pred  = reg_model.predict(X_test)
np.sqrt(mean_squared_error(y_pred, y_test))
```

```python
# MAE Test
np.sqrt(mean_absolute_error(y_test, y_pred))
```

```python
# R² Test
reg_model.score(X_test, y_test)
```

## Validation

```python
# 10 Fold Cross Validation
np.mean(np.sqrt(-cross_val_score(reg_model,
                X,
                y,
                cv=10,
                scoring="neg_mean_squared_error")))
```
